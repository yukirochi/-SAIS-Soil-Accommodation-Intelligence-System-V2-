```python
import pandas as pd

df = pd.read_csv('dataset/soil_fertility_clean (1).csv')

print(df.shape)
df = df.drop(columns=['fertility_class'])
print(df.columns)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

correlation = df.corr(method='spearman')

plt.figure(figsize=(12, 8))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

plt.title('Correlation Heatmap of Soil Fertility Dataset', fontsize=16)
plt.show()
df.describe()
```
